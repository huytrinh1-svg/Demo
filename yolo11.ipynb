```python
from ultralytics import YOLO
import cv2
import matplotlib.pyplot as plt
from PIL import Image
import numpy as np
%matplotlib inline

# Load YOLOv11n
model = YOLO("yolo11n.pt")
print("✅ Model loaded!")
```

```python
print("\n=== KIẾN TRÚC YOLOv11n ===")
print(model.model)
print(f"\nTotal params: {sum(p.numel() for p in model.model.parameters()):,}")
```

```python
image_path = r"C:\Users\huy.trinh1\Downloads\unnamed (1).jpg"
results_all = model(image_path, conf=0.5, verbose=True)

# Hiển thị
plt.figure(figsize=(12, 8))
annotated_rgb = results_all[0].plot()[:, :, ::-1]
plt.imshow(annotated_rgb)
plt.axis('off')
plt.show()


```
